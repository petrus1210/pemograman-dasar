```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import pandas as pd

city_names = ['Jakarta', 'Bandung', 'Makassar', 'Surabaya', 'Medan', 'Yogyakarta', 'Malang']
population = [498044, 964254, 491918, 8398748, 653115, 883305, 744955]
num_airports = [2, 2, 8, 3, 1, 3, 2]

df = pd.DataFrame({
    'City Name': city_names,
    'Population': population,
    'Airports': num_airports,
})

df
```

```python
# Ringkasan statistik
df.describe()

# Rata-rata populasi
rata_rata_populasi = df['Population'].mean()
print("Rata-rata populasi:", rata_rata_populasi)

# Modus jumlah bandara
modus_airports = df['Airports'].mode()
print("Modus Airports:", modus_airports)
```

```python
import matplotlib.pyplot as plt

df['Population'].hist()
plt.title("Distribusi Populasi Kota")
plt.xlabel("Jumlah Penduduk")
plt.ylabel("Frekuensi")
plt.show()
```

```python
# Korelasi
corr_matrix = df[['Population', 'Airports']].corr()
print("Korelasi:", corr_matrix)

# Kovarian
cov_matrix = df[['Population', 'Airports']].cov()
print("Kovarian:", cov_matrix)
```

```python
# Menambah kolom kategori
df['Kategori_Bandara'] = df['Airports'].apply(lambda x: 'Banyak' if x >= 3 else 'Sedikit')
df

# Rata-rata populasi per kategori
group_mean = df.groupby('Kategori_Bandara')['Population'].mean()
print(group_mean)

# Agregasi lengkap
group_agg = df.groupby('Kategori_Bandara').agg({
    'Population': ['mean', 'max'],
    'Airports': ['mean', 'count']
})
group_agg
```

PEJELASAN TENTANG LEANING DAN EDA PYTHON

1.Pengertian data preparation Data preparation adalah proses pengolahan data mentah menjadi data yang siap digunakan untuk analisis atau pemprosesan lebih lanjut. Proses ini melibatkan beberapa Langkah, seperti:

Pengumpulan data : mengumpulkan data dari berbagai sumber, seperti data base, file atau sumber lainnya.

Pembersihan data : membersihkan data dari kesalahan

Transformasi data : mengubah format data menjadi format yang lebih sesuai untuk analisis

Integrasi data : menggabungkan data dari berbagai sumber menjadi satu data set yang terintegrasi

Pengurangan data : mengurangi ukuran data untuk mempercepat proses analisis atau pemrosesan

Pengaturan data : mengatur data menjadi format yang lebih mudah digunakan. Ibarat membuat sup : belanja – mencuci – memotong – memasak – sup siap Begitu juga dengn data : kumpulkan – bersihkan – periksa – siap di analisis

2.Variabel/peubah Variable/peubah adalah sebuah karakteristik yang dapat mengandung berbagai nilai dari objek yang diamati. Variabel/peubah dapat dibedakan menjadi dua bagian yaitu variabel kategorik dan numerik. Contoh dari variabel : - Jenis kelamin : pria atau Wanita

Usia : 5, 10, 50…., 60 tahun
Pendidikan : SD, SMP, SMA,KULIAH ….

3.Jenis-Jenis Data

Data Cross Section Data yang terdiri dari banyak objek yang diukur dalam satu waktu yang sama contoh : GDP dari 100 pegora di dunia tahun 2024
Data Time Series Data yang hanya terdiri dari satu objek yang diukur banyak titik waktu contoh : GDP Indonesia tahun 2000-2024
Data Panel Data yang terdiri dari banyak objek yang diukur di banyak titik waktu, contoh : GDP dari 100 negara di dunia tahun 2000-2024
Data Hilang (missing value)

4.Data hilang merupakan suatu kejadian Ketika tidak ada nilai data yang terekam/tercatat pada pengamatan untuk suatu variabel Penyebab data hilang :

Non – respon di survei
Kondisi pengamatan rusak
Kesalahan peneliti atau entri data.

5.Jenis Mekanisme Data Hilang

Missing completely at Random (MCAR)
Missing at Random (MAR)
Missing Not at Random (MNAR)

6.Implikasi Data Hilang dalam Analisis

Ketiadaan data pengamatan akan mengurangi kekuatan statistic
Adanya data yang hilang dapat meningkatkan galat baku (standard error)
Data yang hilang dapat menyebabkan bias dalam pendugaan parameter
Dapat mengurangi keterwakilan sampel sehingga melemahkan generalisasi dari hasil
Adanya data hilang dapat mempersulit analisis penelitian

7.Metode-metode mengatasi Data yang Hilang

Penghapusan kasus (Listuise or case peletion)
Penghapusan berpasangan (pair wise detection)
Penggantian dengan rataan (Mean subtition)
Pengisian dengan regresi (Regression Imputation)

8.Pencilan (outlier) Pencilan adalah titik data dimana nilainya jauh dari nilai secara umum (yang di prediksi oleh model)
